## Baca CSV

Sel ini menyimpan data kecil sebagai CSV di sesi notebook, lalu membacanya dengan pandas. Pada latihan mandiri, unggah CSV lain melalui panel Files.

```python
from pathlib import Path
import pandas as pd
CSV_TEXT = 'moda,waktu_menit\nBus,12\nBus,18\nMotor,10\nMotor,14\nJalan kaki,8\nJalan kaki,16\nSepeda,12\nSepeda,30\n'
Path('perjalanan_contoh.csv').write_text(CSV_TEXT, encoding='utf-8')
data = pd.read_csv('perjalanan_contoh.csv')
print(data.shape)
print(data.to_string(index=False))
```

## Periksa kualitas awal

Nol nilai kosong tidak membuktikan data ini mewakili suatu kota.

```python
print(data.isna().sum().to_string())
print('Baris duplikat:', data.duplicated().sum())
```

## Ringkas kategori dan waktu

Coba hitung rata-rata dengan tangan: jumlah delapan waktu dibagi delapan.

```python
waktu = data['waktu_menit']
print(data['moda'].value_counts().to_string())
print('Rata-rata:', waktu.mean())
print('Median:', waktu.median())
print('Simpangan baku sampel:', round(waktu.std(), 2))
```

## Peluang empiris

A adalah Bus. B adalah waktu lebih dari 15 menit. Sebutkan penyebut sebelum menghitung peluang bersyarat.

```python
A = data['moda'] == 'Bus'
B = data['waktu_menit'] > 15
print('A, B, A dan B:', A.sum(), B.sum(), (A & B).sum())
print('P(B | A):', (A & B).sum() / A.sum())
```

## Coba sendiri

Ubah batas waktu dari `> 15` menjadi `> 20`. Prediksi banyak baris yang terpilih. Tulis kesimpulan yang menyebut bahwa data ini simulasi.
